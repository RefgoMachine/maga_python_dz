1840e0b9d4 ‘Продукты‘

# Задание 2

Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Если покупка была, сам файл visit_log.csv изменять не надо.

Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.

Учтите условия для данных:

- содержимое purchase_log.txt помещается в оперативную память компьютера;
- содержимое visit_log.csv — нет, используйте только построчную обработку этого файла.

 
#### Примеры работы программы при выводе первых трёх строк файла funnel.csv:

```
user_id,source,category

1840e0b9d4,other,Продукты

4e4f90fcfb,context,Электроника
```

````python
import json
import csv

dict_category_with_user = {}
funnel_csv = []
funnel_csv_fieldnames = ["user_id", "source", "category"]

with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    for line in f.readlines():
        json_log = json.loads(line.strip())
        dict_category_with_user[json_log["user_id"]] = json_log["category"]

with open('visit_log.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        row["category"] = dict_category_with_user.get(row["user_id"])
        if row["category"] != None:
            funnel_csv.append(row)

with open('funnel.csv', 'w') as f:
    writer = csv.DictWriter(f, fieldnames=funnel_csv_fieldnames)
    writer.writeheader()
    writer.writerows(funnel_csv)

with open('funnel.csv', 'r', encoding='utf-8') as f:
    for line in f.readlines()[0:10:]:
        print(line)
````
